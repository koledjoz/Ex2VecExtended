```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import pearsonr
```

```python
df = pd.read_csv('../analysis/sim_analysis/original.csv')

correlations = []

for s in df['split'].unique():
    tmp_df = df[df['split'] == s]
    correlations.append(pearsonr(tmp_df['prob'], tmp_df['item_dist']).statistic)

print('ORIGINAL MODEL')
print(f'{np.mean(correlations):.4f} ± {np.std(correlations):.4f}')
```

```python
df = pd.read_csv('../analysis/sim_analysis/extended.csv')

correlations = []

for s in df['split'].unique():
    tmp_df = df[df['split'] == s]
    correlations.append(pearsonr(tmp_df['prob'], tmp_df['item_dist']).statistic)

print('EXTENDED MODEL')
print(f'{np.mean(correlations):.4f} ± {np.std(correlations):.4f}')
```

```python
df = pd.read_csv('../analysis/sim_analysis/extendedDoubleBase.csv')

correlations_base = []
correlations_ext = []

for s in df['split'].unique():
    tmp_df = df[df['split'] == s]
    correlations_base.append(pearsonr(tmp_df['prob'], tmp_df['item_dist_base']).statistic)
    correlations_ext.append(pearsonr(tmp_df['prob'], tmp_df['item_dist_ext']).statistic)

print('EXTENDED DOUBLE BASE MODEL')
print(f'Base embedding: {np.mean(correlations_base):.4f} ± {np.std(correlations_base):.4f}')
print(f'Extended embedding: {np.mean(correlations_ext):.4f} ± {np.std(correlations_ext):.4f}')
```
