```python
import pandas as pd
import random

from tqdm import tqdm

import json
```

```python
df = pd.read_parquet('./sorted_data.parquet')
```

```python
df[['user_id', 'track_id']].drop_duplicates()['user_id'].value_counts()
```

```python
def split_set(data, train_ratio=0.7, test_ratio=0.2, val_ratio=0.1):
    if not isinstance(data, set):
        raise TypeError("Input must be a set.")
    if len(data) < 3:
        raise ValueError("Set must contain at least 3 elements to split into train, test, and validation sets.")

    data_list = list(data)
    random.shuffle(data_list)

    # Calculate sizes, ensuring each split gets at least one item
    total = len(data_list)
    train_size = max(1, int(total * train_ratio))
    test_size = max(1, int(total * test_ratio))
    val_size = total - train_size - test_size

    # Adjust if rounding causes issues
    if val_size < 1:
        val_size = 1
        if train_size > test_size:
            train_size -= 1
        else:
            test_size -= 1

    train_set = set(data_list[:train_size])
    test_set = set(data_list[train_size:train_size + test_size])
    val_set = set(data_list[train_size + test_size:])

    return train_set, test_set, val_set
```

```python
train_dict = {}
test_dict = {}
val_dict = {}

for user in tqdm(df['user_id'].unique()):
    track_set = set([int(x) for x in df[df['user_id'] == user]['track_id'].unique()])

    train_set, test_set, val_set = split_set(track_set, train_ratio=0.8, test_ratio=0.1, val_ratio=0.1)

    train_dict[int(user)] = list(train_set)
    test_dict[int(user)] = list(test_set)
    val_dict[int(user)] = list(val_set)
```

```python
with open('train_dict.json', "w") as f:
        json.dump(train_dict, f, indent=4)

with open('test_dict.json', "w") as f:
        json.dump(test_dict, f, indent=4)

with open('val_dict.json', "w") as f:
        json.dump(val_dict, f, indent=4)
```
